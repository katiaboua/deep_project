```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import time
import sys
sys.path.append('..')

from environments.grid_world import GridWorldEnv
from algorithms.dynamic_programming import (
    iterative_policy_evaluation,
    policy_iteration,
    value_iteration
)

print("Imports OK ✅")
```

```python
S = GridWorldEnv.S
A = GridWorldEnv.A
R = GridWorldEnv.R
T = GridWorldEnv.T
p = GridWorldEnv.p

print("MDP GridWorld défini ✅")
print(f"États : {len(S)} états (grille 5x5)")
print(f"Actions : {A} (0=Haut, 1=Bas, 2=Gauche, 3=Droite)")
print(f"Rewards : {R}")
print(f"États terminaux : {T} (4=Trap, 24=Goal)")
```

```python
# Policy Iteration
t1 = time.time()
pi_pi, V_pi = policy_iteration(S, A, R, p, T, gamma=0.99, theta=0.001)
t1 = time.time() - t1

# Value Iteration
t2 = time.time()
pi_vi, V_vi = value_iteration(S, A, R, p, T, gamma=0.99, theta=0.001)
t2 = time.time() - t2

print("=== Policy Iteration ===")
print(V_pi.reshape(5,5).round(3))
print(f"Temps : {t1:.4f}s")

print("\n=== Value Iteration ===")
print(V_vi.reshape(5,5).round(3))
print(f"Temps : {t2:.4f}s")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, V, title in zip(axes,
                         [V_pi, V_vi],
                         ['Policy Iteration', 'Value Iteration']):
    im = ax.imshow(V.reshape(5,5), cmap='RdYlGn', vmin=-3, vmax=1)
    plt.colorbar(im, ax=ax)

    for i in range(5):
        for j in range(5):
            s = i * 5 + j
            if s == 4:
                ax.text(j, i, 'TRAP', ha='center', va='center', color='black', fontsize=8)
            elif s == 24:
                ax.text(j, i, 'GOAL', ha='center', va='center', color='black', fontsize=8)
            else:
                ax.text(j, i, f'{V[s]:.2f}', ha='center', va='center', color='black', fontsize=8)

    ax.set_title(f'Value Function — {title}')
    ax.set_xticks([])
    ax.set_yticks([])

plt.suptitle('Comparaison V(s) — GridWorld', fontsize=14)
plt.tight_layout()
plt.savefig('gridworld_V.png', dpi=100, bbox_inches='tight')
plt.close()
print("Graphique sauvegardé ✅")
```

```python
ACTION_ARROWS = {0: '↑', 1: '↓', 2: '←', 3: '→'}

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, pi, title in zip(axes,
                          [pi_pi, pi_vi],
                          ['Policy Iteration', 'Value Iteration']):
    ax.set_xlim(-0.5, 4.5)
    ax.set_ylim(-0.5, 4.5)
    ax.set_xticks(range(5))
    ax.set_yticks(range(5))
    ax.grid(True)
    ax.invert_yaxis()

    for i in range(5):
        for j in range(5):
            s = i * 5 + j
            if s == 4:
                ax.add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1, color='red', alpha=0.5))
                ax.text(j, i, 'TRAP', ha='center', va='center', fontsize=8)
            elif s == 24:
                ax.add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1, color='green', alpha=0.5))
                ax.text(j, i, 'GOAL', ha='center', va='center', fontsize=8)
            else:
                best_a = np.argmax(pi[s])
                ax.text(j, i, ACTION_ARROWS[best_a], ha='center', va='center', fontsize=16)

    ax.set_title(f'Policy — {title}')

plt.suptitle('Comparaison Policy — GridWorld', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
gammas = [0.1, 0.5, 0.9, 0.99, 0.999]

fig, axes = plt.subplots(1, len(gammas), figsize=(20, 4))

for ax, g in zip(axes, gammas):
    _, V = value_iteration(S, A, R, p, T, gamma=g, theta=0.001)
    im = ax.imshow(V.reshape(5,5), cmap='RdYlGn', vmin=-3, vmax=1)
    for i in range(5):
        for j in range(5):
            s = i * 5 + j
            if s == 4:
                ax.text(j, i, 'X', ha='center', va='center', fontsize=10)
            elif s == 24:
                ax.text(j, i, 'G', ha='center', va='center', fontsize=10)
            else:
                ax.text(j, i, f'{V[s]:.1f}', ha='center', va='center', fontsize=7)
    ax.set_title(f'gamma={g}')
    ax.set_xticks([])
    ax.set_yticks([])

plt.suptitle('Impact de gamma sur V — GridWorld', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
thetas = [0.1, 0.01, 0.001, 0.0001, 0.00001]
_, V_ref = value_iteration(S, A, R, p, T, gamma=0.99, theta=0.000001)

times = []
precisions = []

for theta in thetas:
    t = time.time()
    _, V = value_iteration(S, A, R, p, T, gamma=0.99, theta=theta)
    times.append(time.time() - t)
    precisions.append(np.mean(np.abs(V - V_ref)))
    print(f"theta={theta:.5f} | temps={times[-1]:.4f}s | précision={precisions[-1]:.6f}")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(thetas, times, marker='o', color='blue')
ax1.set_xlabel('Theta')
ax1.set_ylabel('Temps (s)')
ax1.set_title('Impact de theta sur le temps — GridWorld')
ax1.set_xscale('log')

ax2.plot(thetas, precisions, marker='o', color='red')
ax2.set_xlabel('Theta')
ax2.set_ylabel('Erreur moyenne')
ax2.set_title('Impact de theta sur la précision — GridWorld')
ax2.set_xscale('log')

plt.tight_layout()
plt.show()
```

```python
def value_iteration_track(model_S, model_A, model_R, model_p, model_T, gamma=0.99, theta=0.001):
    V = np.random.random(len(model_S))
    V[model_T] = 0.0
    V_history = [V.copy()]

    while True:
        delta = 0.0
        for s_index in range(len(model_S)):
            v = V[s_index]
            best_value = None
            for a_index in range(len(model_A)):
                total = 0.0
                for s_p_index in range(len(model_S)):
                    for r_index in range(len(model_R)):
                        total += model_p[s_index, a_index, s_p_index, r_index] * (
                            model_R[r_index] + gamma * V[s_p_index]
                        )
                if best_value is None or total > best_value:
                    best_value = total
            V[s_index] = best_value
            delta = np.maximum(delta, np.abs(best_value - v))
        V_history.append(V.copy())
        if delta < theta:
            break

    return V, V_history

V_final, V_history = value_iteration_track(S, A, R, p, T)
print(f"Convergence en {len(V_history)} itérations")

# Afficher quelques itérations clés
iterations_to_show = [0, 1, 2, len(V_history)//2, len(V_history)-1]
fig, axes = plt.subplots(1, len(iterations_to_show), figsize=(20, 4))

for ax, i in zip(axes, iterations_to_show):
    im = ax.imshow(V_history[i].reshape(5,5), cmap='RdYlGn', vmin=-1, vmax=1)
    for row in range(5):
        for col in range(5):
            s = row * 5 + col
            ax.text(col, row, f'{V_history[i][s]:.2f}',
                   ha='center', va='center', fontsize=7)
    ax.set_title(f'Itération {i}')
    ax.set_xticks([])
    ax.set_yticks([])

plt.suptitle(f'Propagation de V — GridWorld ({len(V_history)} itérations)', fontsize=14)
plt.tight_layout()
plt.show()
```
